## 1. `print()` y operaciones

`print()` es nuestra primera instrucción. Muestra un valor en pantalla.  
Python también sabe hacer cuentas como una calculadora.

| Operador | Significado | Ejemplo |
|:--------:|-------------|--------:|
| `+ -` | Suma, resta | `1250 + 50` |
| `* /` | Producto, división | `1250 * 72.50` |
| `**` | Potencia | `2 ** 10` |
| `//` | División entera | `1250 // 30` |
| `%` | Residuo (módulo) | `1250 % 30` |

Python respeta el orden matemático: primero `*` y `/`, después `+` y `-`.

````python
print("Hola SLB Ecuador")
print(1250 * 72.50)              # ingreso bruto: BOPD * precio
print(1250 * (72.50 - 38.00))    # utilidad diaria: BOPD * (precio - costo)
print(1250 ** 2)                 # potencia
````

## 2. Práctica guiada: Rentabilidad de un pozo

> El pozo **Sacha-042** produce **1 250 barriles/día (BOPD)**.  
> El crudo se vende a **\$72.50/bbl** y el costo operativo es **\$38.00/bbl**.

Usando solo `print()` y operaciones, calcula:

1. El **ingreso bruto diario** del pozo.
2. La **utilidad diaria** (ingreso menos costo).
3. El **margen operativo en porcentaje**.

> 💡 Pista: margen = (precio − costo) / precio × 100

````python
# Solución guiada
print(1250 * 72.50)                          # ingreso bruto
print(1250 * 72.50 - 1250 * 38.00)           # utilidad diaria
print((72.50 - 38.00) / 72.50 * 100)         # margen operativo %
````

## 3. Variables: asignar y reutilizar

Una **variable** es un nombre que apunta a un valor en memoria.

- `=` **no** significa "igual a". Significa **asignar**: el valor de la derecha se guarda con el nombre de la izquierda.
- Si el precio del crudo sube a \$78, cambiamos **una sola línea** y todo se recalcula.
- `Precio` ≠ `precio`. Python distingue mayúsculas.

````python
bopd   = 1250
precio = 72.50
costo  = 38.00

ingreso  = bopd * precio
utilidad = ingreso - bopd * costo

print("Ingreso:", ingreso)
print("Utilidad:", utilidad)
````

### Nombrar variables importa

Un buen nombre hace que el código se explique solo.

| ✅ Correcto | ❌ Evitar |
|-------------|----------|
| `bopd_pozo` | `b` |
| `precio_crudo` | `p1` |
| `corte_agua_pct` | `dato` |
| `costo_op_bbl` | `x` |

**Convención Python**: `snake_case` — palabras en minúscula separadas por `_`.

## 4. Tipos de datos

Python clasifica cada valor. El **tipo** determina qué operaciones son válidas.

| Tipo | Ejemplo | Uso en operación |
|------|---------|------------------|
| `int` — entero | `1250` | Barriles, número de pozos |
| `float` — decimal | `72.50` | Precios, cortes de agua, porosidad |
| `str` — texto | `"Sacha-042"` | Nombre de pozo, campo, formación |
| `bool` — lógico | `True / False` | ¿Pozo activo? ¿Supera cutoff? |

````python
print(type(1250))
print(type(72.50))
print(type("Sacha-042"))
print(type(True))
````

### Cuidado al mezclar tipos

Sumar texto con número no funciona directamente. Python nos avisa con un `TypeError`.

`str()`, `int()` y `float()` **convierten** entre tipos.

````python
# Esto falla:
# print("BOPD: " + 1250)   # TypeError: can only concatenate str (not "int") to str

# Dos soluciones:
print("BOPD: " + str(1250))   # convertimos el número a texto
print("BOPD:", 1250)          # la coma en print acepta cualquier tipo
````

## 5. `input()` — leer datos del usuario

**`input()` pausa el programa y espera a que el usuario escriba.**  
Lo que el usuario escribe se guarda en una variable como texto (`str`).

```
  input("Pozo: ")  →  [pausa]  →  usuario escribe: Sacha-042  →  devuelve "Sacha-042"
```

Puntos clave:

- El texto entre paréntesis es el **mensaje** que ve el usuario.
- Lo escrito se guarda en la variable como **cadena de texto (`str`)**.
- ⚠️ Aunque el usuario escriba `1250`, Python lo guarda como el **texto** `"1250"`, **no** como número.

> 💡 En Colab, al ejecutar la celda aparece un cuadro de texto abajo esperando tu entrada.

````python
# Ejemplo básico: pedir el nombre del pozo
pozo = input("Pozo: ")
print(pozo)
print(type(pozo))   # siempre es <class 'str'>
````

### `input()` + conversión: para operar con números

Si el dato es numérico, **envolvemos** `input()` con `int()` o `float()`:

- `int(input(...))` se lee **de adentro hacia afuera**: primero `input()` pide el dato, después `int()` lo convierte.
- `int()` para enteros (barriles, número de pozos).
- `float()` para decimales (precio, corte de agua).

````python
pozo   = input("Pozo: ")
bopd   = int(input("BOPD: "))
precio = float(input("Precio ($/bbl): "))

ingreso = bopd * precio
print(pozo, "-> ingreso: $", ingreso)
````

## 6. Comparaciones y booleanos

Toda comparación devuelve `True` o `False`. Con eso el programa decide.

| Operador | Significado |
|:--------:|-------------|
| `==` | Igual a |
| `!=` | Distinto de |
| `>` `<` | Mayor / menor |
| `>=` `<=` | Mayor-igual / menor-igual |
| `and` | Ambas ciertas |
| `or` | Al menos una cierta |
| `not` | Niega |

````python
bopd = 1250
corte_agua = 0.32

print(bopd > 1000)
print(corte_agua < 0.5)
print(bopd > 1000 and corte_agua < 0.5)
print(bopd < 500 or corte_agua > 0.9)
````

## 7. Condicionales: `if / elif / else`

El programa toma un camino según el dato. La **indentación** (4 espacios) define el bloque.

- El `:` al final de la línea abre un bloque.
- `elif` se puede repetir tantas veces como quieras.
- `else` es opcional.
- Solo se ejecuta **una** rama: la primera que sea `True`.

````python
bopd = 1250

if bopd >= 1000:
    estado = "productor alto"
elif bopd >= 300:
    estado = "productor medio"
else:
    estado = "candidato a workover"

print("Sacha-042:", estado)
````

## 8. Listas: varios valores, un nombre

Cuando no tenemos un pozo, sino una **batería de pozos**.

- Se crean con `[ ]` separando por comas.
- Se indexan desde **0**. `-1` es el último.
- `lista[a:b]` devuelve del `a` al `b-1`: **slice**.
- `len()` cuenta, `sum()` suma, `append()` agrega al final.
- **Mismo orden** entre listas paralelas: la posición 0 es el mismo pozo en ambas.

````python
pozos = ["Sacha-042", "Sacha-108", "Shushufindi-21", "Auca-63"]
bopd  = [1250, 780, 1600, 340]

print(pozos[0])       # primer pozo
print(bopd[-1])       # ultimo BOPD
print(pozos[1:3])     # slice: del indice 1 al 2
print(len(bopd))      # cuantos hay

bopd.append(920)      # agregar al final
print(sum(bopd))      # suma total de la produccion
````

---
## 9. Ciclos: `for` — repetir sin copiar-pegar

`for` recorre una lista y ejecuta el mismo bloque para cada elemento.

`zip()` empareja dos listas posición por posición para recorrerlas en paralelo.

````python
pozos = ["Sacha-042", "Sacha-108", "Shushufindi-21", "Auca-63"]
bopd  = [1250, 780, 1600, 340]

for produccion in bopd:
    print(produccion, "BOPD")
````

````python
# Recorrer dos listas en paralelo con zip()
for pozo, prod in zip(pozos, bopd):
    print(pozo, "->", prod)
````

### `for` + `if`: filtrar y contar

¿Cuántos pozos superan el cutoff de **1 000 BOPD**? ¿Cuánto produce la batería en total?

Este es el patrón **acumulador + contador**: la base de lo que después haremos en `pandas` con una sola línea.

````python
pozos = ["Sacha-042", "Sacha-108", "Shushufindi-21", "Auca-63"]
bopd  = [1250, 780, 1600, 340]

sobre_cutoff = 0
total = 0

for pozo, prod in zip(pozos, bopd):
    total = total + prod
    if prod >= 1000:
        sobre_cutoff = sobre_cutoff + 1
        print(pozo, "supera cutoff")

print("Total:", total, "BOPD")
print("Sobre cutoff:", sobre_cutoff)
````

## 10. Caso Integrador: Reporte diario del pozo Sacha-042

> Antes del handover de las 7 PM, ingeniería de producción necesita un **reporte diario** del pozo **Sacha-042** para el jefe de campo.

**Datos del pozo:**

- **Producción de líquido total**: 1 840 barriles/día
- **Corte de agua**: 32 % (0.32)
- **Precio del crudo Oriente**: \$72.50 / bbl
- **Costo operativo (lifting cost)**: \$38.00 / bbl de líquido
- **Cutoff de rentabilidad**: 500 BOPD netos

**El reporte debe entregar:**

1. **BOPD netos** (líquido total × (1 − corte_agua)) y **BWPD** (agua).
2. **Ingreso, costo, utilidad diaria y margen %**.
3. **Clasificación del pozo** con `if / elif / else`:
   - Si `bopd_netos >= 1000` **y** `corte_agua < 0.5` → **"óptimo"**
   - Si `corte_agua >= 0.9` → **"candidato a cierre"**
   - Si `bopd_netos >= cutoff` → **"rentable"**
   - En cualquier otro caso → **"revisión técnica"**
4. **Proyección mensual** de utilidad (30 días).

> 💡 El objetivo no es solo calcular: es producir un resumen que el jefe de campo pueda leer en 30 segundos.

````python
# Espacio para tu solución
````

### Solución sugerida

Corre esta celda solo después de intentar la tuya.

````python
# Datos del pozo
pozo         = "Sacha-042"
liquido_bpd  = 1840
corte_agua   = 0.32
precio_crudo = 72.50
costo_op_bbl = 38.00
cutoff       = 500

# 1. BOPD netos y BWPD
bopd_netos = liquido_bpd * (1 - corte_agua)
bwpd       = liquido_bpd * corte_agua

# 2. Métricas financieras diarias
ingreso  = bopd_netos * precio_crudo
costo    = liquido_bpd * costo_op_bbl
utilidad = ingreso - costo
margen   = utilidad / ingreso * 100

# 3. Clasificación del pozo
if bopd_netos >= 1000 and corte_agua < 0.5:
    estado = "optimo"
elif corte_agua >= 0.9:
    estado = "candidato a cierre"
elif bopd_netos >= cutoff:
    estado = "rentable"
else:
    estado = "revision tecnica"

# 4. Proyección mensual (30 días)
utilidad_mes = utilidad * 30

# Reporte
print("=" * 42)
print(" REPORTE DIARIO -", pozo)
print("=" * 42)
print("BOPD netos:      ", round(bopd_netos, 1))
print("BWPD:            ", round(bwpd, 1))
print("-" * 42)
print("Ingreso:         $", round(ingreso, 2))
print("Costo op.:       $", round(costo, 2))
print("Utilidad diaria: $", round(utilidad, 2))
print("Margen:          ", round(margen, 1), "%")
print("-" * 42)
print("Estado:          ", estado)
print("Proyeccion 30d:  $", round(utilidad_mes, 2))
print("=" * 42)
````
