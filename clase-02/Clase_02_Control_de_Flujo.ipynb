# Parte 1 · Condicionales a fondo

## `if` — anatomía

Un `if` evalúa una condición. Si es `True`, ejecuta el bloque indentado; si es `False`, lo salta.

- `if` + condición + `:`
- El **bloque** son las líneas con **4 espacios** de indentación.
- Lo que **no** está indentado se ejecuta siempre.

```python
presion = 3200

if presion > 3000:
    print("Alerta: presion alta")

print("Fin")
```

## `if` solo — "hacer algo solo si..."

Se usa cuando **no hay caso alternativo**: solo actuamos si pasa algo especial. Como la presión es 2800, la condición es `False` y no se imprime nada.

```python
presion = 2800

if presion > 3000:
    print("Alerta: presion alta")
```

## `if / else` — dos caminos exclusivos

Se ejecuta **exactamente una** de las dos ramas. `else` no lleva condición: es "todos los demás casos".

```python
bopd = 750

if bopd >= 500:
    estado = "activo"
else:
    estado = "cerrado"

print("Pozo:", estado)
```

## `if / elif / else` — cadena de casos

Python evalúa **de arriba abajo** y ejecuta **solo el primero** que sea `True`; luego sale de la cadena.
`elif` (= "else if") se puede repetir tantas veces como haga falta.

```python
bopd = 1250

if bopd >= 1000:
    nivel = "alto"
elif bopd >= 300:
    nivel = "medio"
else:
    nivel = "workover"

print(nivel)
```

## Trampa: `if / if / if` ≠ `if / elif / elif`

Tres `if` **sueltos** son independientes: se evalúan por separado y pueden imprimir varias cosas. Una **cadena** `if / elif` imprime solo una. Compara las dos celdas:

```python
# Antipatron: tres if sueltos -> imprime las TRES etiquetas
bopd = 1500

if bopd >= 300:
    print("medio")
if bopd >= 800:
    print("alto")
if bopd >= 1200:
    print("muy alto")
```

```python
# Correcto: cadena elif -> imprime SOLO una
bopd = 1500

if bopd >= 1200:
    print("muy alto")
elif bopd >= 800:
    print("alto")
elif bopd >= 300:
    print("medio")
```

## `else` como "catch-all"

`else` atrapa todo lo que no capturaron los casos anteriores. Es útil para **garantizar un valor**, pero cuidado: también puede **esconder errores** (por ejemplo, un error de tipeo cae al `else` en silencio).

```python
bopd = 120

if bopd >= 1000:
    nivel = "alto"
elif bopd >= 300:
    nivel = "medio"
else:
    nivel = "revisar"

print(nivel)
```

## Combinar condiciones: `and`, `or`, `not`

- `and` — deben cumplirse **las dos**.
- `or` — basta con **una**.
- `not` — invierte `True`/`False`.
- Los **paréntesis** agrupan y dejan claro el orden.

(`corte` = corte de agua.)

```python
bopd = 1250
corte = 0.32

print(bopd >= 1000 and corte < 0.5)
print(bopd < 200 or corte > 0.9)
print(not False)
```

# Parte 2 · Listas

## ¿Qué es una lista?

En la Clase 1 teníamos **un** pozo. Hoy el supervisor pide la producción de los **5 pozos** de la batería.

Sin listas necesitaríamos una variable por pozo (`bopd_1`, `bopd_2`, ... — imposible si son 500). Una **lista** guarda muchos valores ordenados bajo un solo nombre:

- Los corchetes `[ ]` delimitan la lista.
- Las comas separan los elementos.
- Es **ordenada**: cada valor tiene una posición fija.

```python
bopd = [1250, 780, 340, 1600, 950]
print(bopd)
```

## Acceder a un elemento: el índice

Cada elemento tiene una **posición** (índice). Python cuenta **desde 0**.

| valor | 1250 | 780 | 340 | 1600 | 950 |
|-------|:----:|:---:|:---:|:----:|:---:|
| índice | 0 | 1 | 2 | 3 | 4 |

- El primer elemento es `[0]`, no `[1]`.
- `[-1]` es el último (cuenta desde el final).
- `len()` dice cuántos elementos hay.

```python
bopd = [1250, 780, 340, 1600, 950]

print(bopd[0])    # primer pozo
print(bopd[2])    # tercero
print(bopd[-1])   # ultimo
print(len(bopd))  # cuantos hay
```

## Slicing: cortar un pedazo de la lista

`lista[a:b]` devuelve desde el índice `a` hasta `b-1` (**incluye `a`, excluye `b`**). Omitir el primer número = "desde el inicio"; omitir el segundo = "hasta el final".

```python
prod = [1250, 1180, 1220, 1150, 1300]

print(prod[0:3])   # indices 0,1,2
print(prod[2:])    # del 2 al final
print(prod[:2])    # los 2 primeros
print(prod[-2:])   # los 2 ultimos
```

## Modificar una lista: los métodos

Un **método** es una acción que se llama con la sintaxis `lista.metodo(...)`.

| Método | Qué hace |
|--------|----------|
| `.append(x)` | agrega `x` al final |
| `.insert(i, x)` | inserta `x` en la posición `i` |
| `.remove(x)` | borra la primera aparición de `x` |
| `.pop()` | quita y devuelve el último |

```python
pozos = ["Sacha-042", "Sacha-108"]

pozos.append("Auca-63")
pozos.insert(0, "Auca-91")
pozos.remove("Sacha-108")

print(pozos)
```

## Operador `in`: ¿está el valor en la lista?

`valor in lista` devuelve `True` si el valor aparece; `False` si no. Sirve para **decidir** sin escribir un ciclo, y combina muy bien con `if`.

```python
ruta = ["Sacha-042", "Auca-63"]

print("Sacha-042" in ruta)
print("Auca-91" in ruta)

if "Sacha-042" in ruta:
    print("visitar hoy")
```

## Listas paralelas: una tabla sin librerías

Dos o más listas donde la **posición** conecta la información. La posición 0 es el mismo pozo en las tres listas.

- Regla de oro: **todas las listas deben tener el mismo largo**.
- En la Clase 3 esto se convierte en un `DataFrame` de pandas.

```python
pozos = ["Sacha-042", "Sacha-108"]
bopd  = [1250, 780]
corte = [0.32, 0.55]

print(pozos[0], bopd[0], corte[0])
```

# Parte 3 · Ciclos `for`

## Anatomía del `for`

`for` **recorre** una lista: para **cada** elemento ejecuta el bloque indentado.

- `prod` es la **variable de bucle**: cambia en cada vuelta.
- `in bopd` le dice a Python **qué recorrer**.
- Solo lo indentado se repite; lo de afuera corre al final.

```python
bopd = [1250, 780, 1600]

for prod in bopd:
    print(prod, "BOPD")

print("fin")
```

## `range(n)`: números para contar

`range(n)` genera **0, 1, ..., n-1**. Útil para repetir un número fijo de veces. `range(a, b)` va de `a` hasta `b-1` (**no incluye** el final).

```python
for dia in range(7):
    print("dia", dia + 1)
```

```python
for h in range(6, 12):
    print("hora", h)
```

## `enumerate()`: índice y valor a la vez

Cuando además del valor necesitamos su **posición**. `i, pozo` son **dos variables** a la vez: `i` la posición, `pozo` el elemento. Arranca en 0 (para empezar en 1: `enumerate(pozos, 1)`).

```python
pozos = ["Sacha-042", "Auca-63"]

for i, pozo in enumerate(pozos):
    print(i, "->", pozo)
```

## `zip()`: recorrer dos listas en paralelo

`zip(a, b)` empareja los elementos de `a` y `b` posición por posición. Es la forma limpia de recorrer **listas paralelas**.

```python
pozos = ["Sacha-042", "Sacha-108"]
bopd  = [1250, 780]

for pozo, prod in zip(pozos, bopd):
    print(pozo, prod)
```

## Patrón acumulador: sumar todo

Empezamos en `total = 0` **antes** del ciclo y en cada vuelta sumamos el elemento actual. Atajo: `sum(prod)` hace lo mismo en una línea.

```python
prod = [1250, 1180, 1220, 1150]

total = 0
for p in prod:
    total = total + p

print("Total:", total)
```

## Patrón contador: ¿cuántos cumplen?

Empezamos en `n = 0` y sumamos `1` cada vez que se cumple una condición. Es un `for` con un `if` **adentro**: revisamos cada pozo por separado.

```python
bopd = [1250, 780, 1600, 340, 920]

n = 0
for prod in bopd:
    if prod >= 1000:
        n = n + 1

print("Sobre cutoff:", n)
```

## Patrón filtro: guardar los que cumplen

Empezamos con una lista **vacía** y hacemos `.append()` de los que pasan el filtro. El `for` recorre, el `if` filtra y `.append()` guarda.

```python
pozos = ["Sacha-042", "Auca-63", "Shushufindi-21"]
bopd  = [1250, 340, 1600]

altos = []
for pozo, prod in zip(pozos, bopd):
    if prod >= 1000:
        altos.append(pozo)

print(altos)
```
